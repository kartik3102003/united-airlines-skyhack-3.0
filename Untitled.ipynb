```python
import os
import pandas as pd
from sqlalchemy import create_engine

# --- MySQL connection details ---
user = "root"
password = "MYSQL0001"
host = "localhost"
database = "united_airlines_db"

# --- Create connection engine ---
engine = create_engine(f"mysql+mysqlconnector://{user}:{password}@{host}/{database}")

# --- Path to your CSV folder ---
folder_path = "/Users/kartikbhardwaj/Desktop/united_airlines"  # ⬅️ CHANGE THIS PATH if needed

# --- Loop through all CSV files in the folder ---
for file in os.listdir(folder_path):
    if file.endswith(".csv"):
        file_path = os.path.join(folder_path, file)

        # Read CSV with UTF-8-SIG encoding to handle BOM issues
        df = pd.read_csv(file_path, encoding='utf-8-sig')

        # Clean table name (remove spaces, +, make lowercase)
        table_name = (
            file.replace(".csv", "")
            .replace(" ", "_")
            .replace("+", "_")
            .lower()
        )

        # Write data to MySQL (replace existing table)
        df.to_sql(name=table_name, con=engine, if_exists='replace', index=False)

        print(f"✅ Imported '{file}' into table '{table_name}' successfully!")

print("\n🎯 All CSV files imported successfully into MySQL database!")
```
